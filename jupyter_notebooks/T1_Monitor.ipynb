import numpy as np
import matplotlib.pyplot as plt


<div style="background-color:#4CAF50; padding:8px 14px; border-radius:4px;">
<h2 style="color:black; margin:0;">Monitor</h2>
</div>

# try/except because there's a brief window where the file could be
# mid-write when you happen to read it — if that happens, just rerun the cell
try:
    with np.load("t1_mwref_checkpoint.npz") as ckpt:
        signal1_sum = ckpt["signal1_sum"]
        signal2_sum = ckpt["signal2_sum"]
        reference1_sum = ckpt["reference1_sum"]
        reference2_sum = ckpt["reference2_sum"]
        t = ckpt["t"]
        passes_done = int(ckpt["passes_done"])

    print(f"{passes_done} passes done so far")

    signal1 = signal1_sum / passes_done
    signal2 = signal2_sum / passes_done

    fig, ax = plt.subplots(figsize=(8, 5))
    ax.plot(t, signal1, 'o-', label='signal1 (no MW)')
    ax.plot(t, signal2, 'o-', label='signal2 (MW pulse)')
    ax.set_xlabel('delay (us)')
    ax.set_ylabel('Counts')
    ax.set_title(f'T1 so far — {passes_done} passes')
    ax.legend()
    plt.show()

except Exception as e:
    print(f"checkpoint busy or not ready, try again: {e}")
